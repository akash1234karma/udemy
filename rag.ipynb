```python
#Data gathering 

from langchain_community.document_loaders import UnstructuredURLLoader, PyPDFLoader
urls=['https://docs.langchain.com/oss/python/langchain/overview']
loader=UnstructuredURLLoader(urls=urls)
docs=loader.load()

#DATA splitting 

from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=75)
split=text_splitter.split_documents(docs)

text=[docs.page_content for docs in split]

#DATA embedding

from langchain_openai import OpenAIEmbeddings
embeddings=OpenAIEmbeddings()
vector=embeddings.embed_documents(text)

# Vector store
from langchain_chroma import Chroma
from langchain_core.documents import Document

vector_store=Chroma.from_documents(documents=split,embedding=OpenAIEmbeddings())
```
