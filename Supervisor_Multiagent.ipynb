```python
from langchain_openai import ChatOpenAI
from langgraph_supervisor import create_supervisor
from langchain.agents import create_agent
from dotenv import load_dotenv
from langchain_community.tools import DuckDuckGoSearchRun

model = ChatOpenAI(model='gpt-4o-mini', temperature=0.4)

def search_duckduckgo(query:str):
    """Search duckduckgo using langchain DuckDuckGoSearchRun tool"""
    search=DuckDuckGoSearchRun()
    return search.invoke(query)

def add(a:int, b:int)-> int:
    """Adding two number a and b"""
    return a + b

def multiply(a:int, b:int)-> int:
    """Multiply two number a and b"""
    return a * b

# create agents

math_agent = create_agent(model=model, tools=[add,multiply], name="maths",
                          system_prompt="You are a math expert, always use one tool at a time")
search_agent = create_agent(model=model,tools=[search_duckduckgo], name='search',
                            system_prompt='you are a world call researcher with access to web search, do not do any math')

# create supervisor

workflow = create_supervisor(
    [math_agent,search_agent],
    model=model,
    prompt=(
        'you are a team supervisor managing a reserch expert and a math expert'
        'for current events, use search_agent'
        'for math problem, use math_agent'
    )
)

app=workflow.compile()
```

```python
result=app.invoke({'messages':[{'role':'user',"content":'Please check Cricbuzz website and see current schedule for Indian cricket team for year 2026'}]})
print(result['messages'][-1].content)

for m in result['messages']:
    m.pretty_print()
```
