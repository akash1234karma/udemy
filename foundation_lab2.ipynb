```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
load_dotenv(override=True)
```

```python
open_api_key=os.getenv('OPENAI_API_KEY')
anthropic_api_key=os.getenv('ANTHROPIC_API_KEY')
google_api_key=os.getenv('GOOGLE_API_KEY')

if open_api_key:
    print(f'openai key exist {open_api_key[:8]}')
else:
    print("openai key not exist")
if anthropic_api_key:
    print(f'anthropic key exist {anthropic_api_key[:8]}')
else:
    print('anthropic key exist')
if google_api_key:
    print(f"Google API Key exists and begins {google_api_key[:2]}")
else:
    print("Google API Key not set (and this is optional)")
```

```python
request = """Please come up with a challenging, nuanced question with a succinct answer,
that I can ask a number of LLMs to evaluate their intelligence.
Not a mathematical puzzle, but more of a thought-provoking question that requires intelligent insight.
Include in your question that the answer must be short."""

request += "Answer only with the question, no explanation"

messages= [{'role': 'user', 'content': request}]
messages
```

```python
openai=OpenAI()

response = openai.chat.completions.create(model='gpt-3.5-turbo', messages=messages)
question=response.choices[0].message.content
display(Markdown(question))
```
