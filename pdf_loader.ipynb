```python
import os
from langchain_community.document_loaders import PyMuPDFLoader,PyPDFLoader,DirectoryLoader
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter

dir_loader=DirectoryLoader("C://Users//a810557//akash//LangchainUpdated//RAG//data", loader_cls=PyMuPDFLoader)
doc=dir_loader.load()



```

```python
def split_documents(documents, chunk_size=500, chunk_overlap=100):
    '''split document into smaller chunk for better RAG performance'''
    text_splitter=RecursiveCharacterTextSplitter(chunk_size=chunk_size,chunk_overlap=chunk_overlap)
    split_docs=text_splitter.split_documents(documents)
    return split_docs
```

```python
def retriver_context(user_query:str, top_k:int=3):
    """Embeds the user query, search ChromaDB, and format retrieved text"""
    query_vector=embeddings_model.encode(user_query).tolist()
    results=collection.query(query_embeddings=[query_vector],n_results=top_k)
    
```
